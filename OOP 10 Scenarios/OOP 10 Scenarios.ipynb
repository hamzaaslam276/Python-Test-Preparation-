and methods to deposit() and withdraw().
* Challenge: Ensure that the balance cannot go negative during withdrawal.


```python
class Bank_Account: 
    def __init__(self,name, balance ): 
        self.name = name
        self.balance = balance
    def deposit(self, deposit_amount):
        self.deposit_amount = deposit_amount
        self.balance += deposit_amount
        print(f"{self.name} Your Deposit Amount {self.deposit_amount} and Current Balance {self.balance}.")
    def withdraw(self,withdraw_amount): 
        self.withdraw_amount = withdraw_amount
        if withdraw_amount < self.balance:
            self.balance -= withdraw_amount
        else:
            print("Insuffiction Balance!")
        print(f"{self.name} Your Withdraw Amount {self.withdraw_amount} and Current Balance {self.balance}.")
hamza = Bank_Account("Hamza", 1000)
hamza.deposit(100)
hamza.withdraw(200)
saleem = Bank_Account("Saleem" , 2000)
saleem.deposit(100)
saleem.withdraw(200)
```

### 3. Book Class for Library System
* Scenario: Implement a Book class with attributes like title, author,
and copies_available. Add a method borrow_book() and return_book() 
that decreases the available copies by 1 if there are any.
* Challenge: Create multiple book objects and track availability.

```python
class Book: 
    def __init__(self, title, author, copies_available): 
        self.title = title
        self.author = author
        self.copies_available = copies_available
    def details(self): 
        print(f"'{self.title}' by {self.author} has {self.copies_available} copies available..")
    def borrow_book(self, book,author_name,borrow_qty):
        self.book = book
        self.author_name = author_name
        self.borrow_qty = borrow_qty
        if book == self.title and author_name == self.author and self.copies_available > 0 and self.copies_available > borrow_qty: 
            self.copies_available -= borrow_qty
            print(f"You have successfully borrowed '{self.title}' by {self.author}.")
        else:
            print(f"Sorry, '{self.title}' by {self.author} is not available for borrowing.")
    def return_book(self,return_book,return_qty):
        if return_qty == self.borrow_qty and return_book == self.book:
            self.copies_available += return_qty
            print(f"Thank you for returning '{self.title}' by {self.author}.")
        elif return_qty != self.borrow_qty: 
            print(f"sorry, Invalid Qty: '{return_qty}'.")
        elif return_book  != self.book:
            print(f"sorry, Invalid Book: '{return_book}'.")
        else:
            print("Sorry Invalid syntax")
b1 = Book("Python", "Hamza", 20)
print(f"Book Title Name: {b1.title}")
print(f"Author Name : {b1.author}")
print(f"Available Qty : {b1.copies_available}")
b1.details()
b1.borrow_book("Python", "Hamza", 2)
b1.return_book("Python", 2)
b1.details()
```

### 5. Shape Class with Polymorphism
* Scenario: Create a base class Shape with a method area(). Then create derived classes Circle, Rectangle, and Triangle that override the area() method to compute their respective areas.
* Challenge: Create a list of shapes and call the area() method on each, demonstrating polymorphism.


```python
class Shape:
    def area(self): 
        return area

class Circle(Shape): 
    def __init__(self,radius,pi = 3.14159): 
        self.radius = radius
        self.pi = pi
    def area(self):
        return self.pi * (self.radius ** 2)
class Rectangle(Shape): 
    def __init__(self, length , width): 
        self.length = length
        self.width = width
    def area(self): 
        return self.length * self.width
class Triangle(Shape): 
    def __init__(self, base, height): 
        self.base = base
        self.height= height
    def area(self):
        return 0.5 *(self.base * self.height)
if __name__ == "__main__": 
    circle = Circle(5)
    rectangle = Rectangle(5,5)
    triangle = Triangle(5,5)

# use for loop
shapes = [
    Circle(radius=7),
    Rectangle(length=7, width=7),
    Triangle(base=7, height=7)
]

# Call the area() method on each shape
for shape in shapes:
    print(f"The area of the {shape.__class__.__name__} is: {shape.area():.2f}")
```

### 7. Library Management System with Composition
* Scenario: Create a Library class that contains a collection of Book objects. Use composition to represent the relationship between the library and its books. The Library class should have methods to add and remove books.
* Challenge: Implement a search feature to find books by title or author.


```python
class Book: 
    def __init__(self, title, author,year): 
        self.title = title
        self.author = author
        self.year = year

    def __str__(self): 
        return f" '{self.title}' by {self.author} ({self.year})."

class Library: 
    def __init__(self):
        self.books = []

    def add(self,book):
        self.books.append(book)
        print(f"Added Book: {book}")
    def remove(self,book): 
        if book in self.books: 
            self.books.remove(book)
            print(f"Removed Book: {book}")
        else:
            print(f"Book not found: {book}")

    def search_books(self,word): 
        results = []
        for book in self.books:
            if word.lower() in book.title.lower() or word.lower() in book.author.lower(): 
                results.append(book)
        if results:
            print(f"Search results for '{word}':")
            for book in results: 
                print(book)
        else:
            print(f"No books found for '{word}'.")

    def display_books(self): 
        if self.books: 
            print("Library Collection:")
            for book in self.books:
                print(book)
        else:
            print("No books in the library.")

if __name__ == "__main__": 
    my_library = Library()
    book1 = Book("Python", "Guido van Rossum", "20, February 1991")
    book2 = Book("Data Science", "Peter Naur", 1974)
    book3 = Book("Machine Learning", "Arthur Samuel", 1959)
    my_library.add(book1)
    my_library.add(book2)
    my_library.add(book3)
    my_library.display_books()
    my_library.search_books("python")
    my_library.remove(book1)
    my_library.display_books()
```

### 8. Movie Rental System with Inheritance and Polymorphism
* Scenario: Create a base class Movie with attributes title, genre, and rating. Derive classes like ComedyMovie, ActionMovie, etc., that override a method get_price() to return different prices based on genre and rating.
* Challenge: Allow rental of movies, track rented movies, and calculate the total rental cost for multiple customers.

```python
# Base class Movie
class Movie:
    def __init__(self, title, genre, rating):
        self.title = title
        self.genre = genre
        self.rating = rating
    
    # This method is meant to be overridden by derived classes
    def get_price(self):
        raise NotImplementedError("This method should be overridden in subclasses")

# Derived class ComedyMovie

class ComedyMovie(Movie):
    def get_price(self):
        if self.rating == "****":
            return 3.5
        else:
            return 2.5

# Derived class ActionMovie
class ActionMovie(Movie):
    def get_price(self):
        if self.rating == "***":
            return 5.0
        else:
            return 4.0

# Derived class DramaMovie
class DramaMovie(Movie):
    def get_price(self):
        return 3.0

# Customer class to track rented movies
class Customer:
    def __init__(self, name):
        self.name = name
        self.rented_movies = []

    def rent_movie(self, movie):
        self.rented_movies.append(movie)

    def calculate_total_cost(self):
        total = 0
        for movie in self.rented_movies:
            total += movie.get_price()
        return total

# MovieRentalStore class to manage multiple customers
class MovieRentalStore:
    def __init__(self):
        self.customers = {}

    def add_customer(self, customer):
        self.customers[customer.name] = customer

    def get_customer_total(self, customer_name):
        if customer_name in self.customers:
            return self.customers[customer_name].calculate_total_cost()
        else:
            return "Customer not found."

# Example Usage
if __name__ == "__main__":
    # Creating movies
    movie1 = ComedyMovie("Funny Movie", "Comedy", "****")
    movie2 = ActionMovie("Explosive Action", "Action", "***")
    movie3 = DramaMovie("Emotional Drama", "Drama", "**")

    # Creating customers
    customer1 = Customer("Saleem")
    customer2 = Customer("Majid")

    # Renting movies
    customer1.rent_movie(movie1)  # Comedy movie
    customer1.rent_movie(movie3)  # Drama movie

    customer2.rent_movie(movie2)  # Action movie

    # Creating the store and adding customers
    store = MovieRentalStore()
    store.add_customer(customer1)
    store.add_customer(customer2)

    # Calculating total rental cost for each customer
    print(f"Total rental cost for {customer1.name}: Rs.{store.get_customer_total('Saleem'):.2f}")
    print(f"Total rental cost for {customer2.name}: Rs.{store.get_customer_total('Majid'):.2f}")
```

### 10. Car Rental System with Advanced Polymorphism and Abstraction
* Scenario: Implement an abstract base class Vehicle with methods like rent() and return_vehicle(). Then create derived classes Car, Truck, and Bike that implement the rental process differently (e.g., price per day, rental policies).
* Challenge: Implement a system where users can rent different types of vehicles, track which vehicles are available or rented, and calculate total rental cost dynamically based on the type of vehicle.

```python
from abc import ABC, abstractmethod

class Vehicle(ABC):
    def __init__(self, license_plate, price_per_day):
        self.license_plate = license_plate
        self.price_per_day = price_per_day
        self.is_rented = False
    
    @abstractmethod
    def rent(self, days):
        """Abstract method to rent the vehicle for a certain number of days"""
        pass

    @abstractmethod
    def return_vehicle(self):
        """Abstract method to return the vehicle"""
        pass

class Car(Vehicle):
    def rent(self, days):
        if not self.is_rented:
            self.is_rented = True
            total_cost = days * self.price_per_day
            print(f"Car {self.license_plate} rented for {days} days. Total cost: Rs.{total_cost}.")
            return total_cost
        else:
            print(f"Car {self.license_plate} is already rented.")
            return 0
    
    def return_vehicle(self):
        if self.is_rented:
            self.is_rented = False
            print(f"Car {self.license_plate} has been returned.")
        else:
            print(f"Car {self.license_plate} was not rented.")


class Truck(Vehicle):
    def rent(self, days):
        if not self.is_rented:
            self.is_rented = True
            total_cost = days * self.price_per_day
            print(f"Truck {self.license_plate} rented for {days} days. Total cost: Rs.{total_cost}.")
            return total_cost
        else:
            print(f"Truck {self.license_plate} is already rented.")
            return 0

    def return_vehicle(self):
        if self.is_rented:
            self.is_rented = False
            print(f"Truck {self.license_plate} has been returned.")
        else:
            print(f"Truck {self.license_plate} was not rented.")

class Bike(Vehicle):
    def rent(self, days):
        if not self.is_rented:
            self.is_rented = True
            total_cost = days * self.price_per_day
            print(f"Bike {self.license_plate} rented for {days} days. Total cost: Rs.{total_cost}.")
            return total_cost
        else:
            print(f"Bike {self.license_plate} is already rented.")
            return 0
    
    def return_vehicle(self):
        if self.is_rented:
            self.is_rented = False
            print(f"Bike {self.license_plate} has been returned.")
        else:
            print(f"Bike {self.license_plate} was not rented.")


class CarRentalSystem:
    def __init__(self):
        self.available_vehicles = []
        self.rented_vehicles = []

    def add_vehicle(self, vehicle):
        self.available_vehicles.append(vehicle)
        print(f"Vehicle {vehicle.license_plate} added to the system.")

    def rent_vehicle(self, license_plate, days):
        for vehicle in self.available_vehicles:
            if vehicle.license_plate == license_plate and not vehicle.is_rented:
                cost = vehicle.rent(days)
                if cost > 0:
                    self.rented_vehicles.append(vehicle)
                    self.available_vehicles.remove(vehicle)
                return
        print(f"Vehicle {license_plate} is not available for rent.")

    def return_vehicle(self, license_plate):
        for vehicle in self.rented_vehicles:
            if vehicle.license_plate == license_plate:
                vehicle.return_vehicle()
                self.available_vehicles.append(vehicle)
                self.rented_vehicles.remove(vehicle)
                return
        print(f"Vehicle {license_plate} is not rented.")

    def show_available_vehicles(self):
        print("\nAvailable Vehicles:")
        for vehicle in self.available_vehicles:
            print(f"{type(vehicle).__name__}: {vehicle.license_plate} - Rs.{vehicle.price_per_day}/day")
        print()

    def show_rented_vehicles(self):
        print("\nRented Vehicles:")
        for vehicle in self.rented_vehicles:
            print(f"{type(vehicle).__name__}: {vehicle.license_plate}")
        print()


rental_system = CarRentalSystem()

# Adding vehicles to the system
car1 = Car("Mahran", 40)
truck1 = Truck("Badford", 80)
bike1 = Bike("CD70", 15)

rental_system.add_vehicle(car1)
rental_system.add_vehicle(truck1)
rental_system.add_vehicle(bike1)

# Show available vehicles
rental_system.show_available_vehicles()

# Rent a vehicle
rental_system.rent_vehicle("Mahran", 3)  # Rent car for 3 days
rental_system.rent_vehicle("CD70", 2)  # Rent bike for 2 days

# Show available and rented vehicles after rental
rental_system.show_available_vehicles()
rental_system.show_rented_vehicles()

# Return vehicles
rental_system.return_vehicle("Mahran")
rental_system.return_vehicle("CD70")

# Show available and rented vehicles after return
rental_system.show_available_vehicles()
rental_system.show_rented_vehicles()
```
